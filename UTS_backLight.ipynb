### Import Library

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

### Import Gambar

```python
img = cv2.imread('Assets/backLiight.jpeg')

if img is None:
    print("Citra tidak dapat dimuat. Pastikan path gambar benar.")
else:
    print("Citra berhasil dimuat.")
```

```python
img.shape
```

```python
[baris, kolom] = img.shape[:2]
```

```python
rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
plt.title('Gambbar Asli')
plt.axis('off')
plt.imshow(rgb)
```

```python
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
plt.imshow(gray, cmap='gray')
plt.title('Gambar Grayscale')
plt.axis('off')
plt.show()
```

```python
gray = cv2.cvtColor(rgb, cv2.COLOR_RGB2GRAY)
baris, kolom = gray.shape

beta = 90
citra_cerah = np.zeros((baris, kolom), dtype=np.uint8)

for x in range(baris):
    for y in range(kolom):
        gyx = int(gray[x, y]) + beta
        gyx = np.clip(gyx, 0, 255)
        citra_cerah[x, y] = gyx

plt.imshow(citra_cerah, cmap='gray')
plt.title('Citra Grayscale Dicerahkan')
plt.axis('off')
plt.show()
```

```python
baris, kolom = gray.shape

alpha = 1.4
citra_kontras = np.zeros((baris, kolom), dtype=np.uint8)

for x in range(baris):
    for y in range(kolom):
        gyx = gray[x, y] * alpha
        gyx = np.clip(gyx, 0, 255)
        citra_kontras[x, y] = gyx

plt.imshow(citra_kontras, cmap='gray')
plt.title('Citra Grayscale DiKontraskan')
plt.axis('off')
plt.show()
```

```python
baris, kolom = citra_cerah.shape

citra_hasil = np.zeros((baris, kolom), dtype=np.uint8)

alpha = 1.2  # >1 menaikkan kontras, <1 menurunkan
beta = 30    # jika ingin menyesuaikan kecerahan tambahan

for i in range(baris):
    for j in range(kolom):
        gyx = gray[i, j] * alpha + beta
        gyx = np.clip(gyx, 0, 255)
        citra_hasil[i, j] = gyx

plt.imshow(citra_hasil, cmap='gray')
plt.title('Citra Grayscale Cerah + Kontras')
plt.axis('off')
plt.show()
```
